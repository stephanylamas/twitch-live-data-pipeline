## Twitch Live Data Pipeline

Project development tutorial and debugging reference.

1. Created file hierarchy for pipeline, data and objects to be stored appropiately
2. Decide on tech stack
3. Read through entire Twitch API Documentations
4. Read through entire EventSub Documentation
5. Create Twitch account 
6. Log into Twitch Developer console https://dev.twitch.tv/
7. Register a Twitch application for the data pipeline
8. Install Twitch CLI

    Project was developed with Windows Subsystem for Linux (WSL). Install homebrew if commands below do not work at https://brew.sh/.

```python
    brew install twitchdev/twitch/twitch-cli 

    brew upgrade twitchdev/twitch/twitch-cli
```

8. Get an OAuth token for application access https://dev.twitch.tv/docs/api/get-started/

```python
curl -X POST 'https://id.twitch.tv/oauth2/token' -H 'Content-Type: application/x-www-form-urlencoded' -d 'client_id=<YOURCLIENTID>&client_secret=<YOURSECRET>&grant_type=client_credentials'
```
